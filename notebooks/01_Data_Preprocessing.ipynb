# Data Cleaning
Load, validate, merge, and export the raw tables.

```python
import pandas as pd
import numpy as np

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
# Path to Excel file
file_path = "../data/raw/EduPro_Online_Platform.xlsx"

# Read sheets
users = pd.read_excel(file_path, sheet_name="Users")
courses = pd.read_excel(file_path, sheet_name="Courses")
transactions = pd.read_excel(file_path, sheet_name="Transactions")
```

Referential Integrity

```python
missing_users = transactions[
    ~transactions["UserID"].isin(users["UserID"])
]

missing_courses = transactions[
    ~transactions["CourseID"].isin(courses["CourseID"])
]

print("Transactions with invalid UserID:", len(missing_users))
print("Transactions with invalid CourseID:", len(missing_courses))
```

Create Age Groups

```python
bins = [0, 17, 25, 35, 45, 100]

labels = [
    "<18",
    "18-25",
    "26-35",
    "36-45",
    "45+"
]

users["AgeGroup"] = pd.cut(
    users["Age"],
    bins=bins,
    labels=labels
)
```

```python
users[["Age", "AgeGroup"]].head()
```

```python
users = users[
    [
        "UserID",
        "UserName",
        "Age",
        "AgeGroup",
        "Gender"
    ]
]
```

```python
courses = courses[
    [
        "CourseID",
        "CourseName",
        "CourseCategory",
        "CourseType",
        "CourseLevel"
    ]
]
```

```python
transactions = transactions[
    [
        "TransactionID",
        "UserID",
        "CourseID",
        "TransactionDate"
    ]
]
```

Merge Datasets

```python
merged = pd.merge(
    transactions,
    users,
    on="UserID",
    how="left"
)
```

```python
merged.head()
```

```python
merged = pd.merge(
    merged,
    courses,
    on="CourseID",
    how="left"
)
```

```python
merged.head()
```

```python
print(merged.shape)

print()

print(merged.isnull().sum())
```

```python
merged.to_csv(
    "../data/processed/merged_data.csv",
    index=False
)

print("Processed dataset saved successfully!")
```

```python
merged["Year"] = merged["TransactionDate"].dt.year
merged["Month"] = merged["TransactionDate"].dt.month
merged["MonthName"] = merged["TransactionDate"].dt.strftime("%B")
merged["Quarter"] = merged["TransactionDate"].dt.quarter
merged["DayOfWeek"] = merged["TransactionDate"].dt.day_name()
```
